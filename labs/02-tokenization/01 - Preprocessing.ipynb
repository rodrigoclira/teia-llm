## Tokenization

*Tokenization* é o processo de dividir um texto de entrada em tokens (palavras ou outros elementos relevantes, como pontuação).

#### Usando expressões regulares

Podemos tokenizar um trecho de texto com um tokenizer baseado em expressão regular, como o disponível no **NLTK**.

Para começar, vamos nos limitar a sequências alfanuméricas de caracteres.

```python
from nltk import regexp_tokenize

text = 'That U.S.A. poster-print costs $12.40...'

pattern = '[a-zA-Z0-9_]+'
tokens = regexp_tokenize(text, pattern)
print(len(tokens))
print(tokens)
```

Podemos refinar a expressão regular para obter uma tokenization mais sensata.

```python
pattern = r'''(?x)           # set flag to allow verbose regexps
        (?:[A-Z]\.)+         # abbreviations, e.g. U.S.A.
        | \w+(?:-\w+)*       # words with optional internal hyphens
        | \$?\d+(?:\.\d+)?%? # currency and percentages, e.g. $12.40, 82%
        | \.\.\.             # ellipsis
        | [][.,;"'?():-_`]   # these are separate tokens; includes ], [
        '''

tokens = regexp_tokenize(text, pattern)
print(len(tokens))
print(tokens)
```

#### Usando o NLTK

O NLTK também inclui um tokenizer de palavras, que chega praticamente ao mesmo resultado (encontra "palavras" e pontuação).

```python
from nltk import word_tokenize

text = 'That U.S.A. poster-print costs $12.40...'
tokens = word_tokenize(text)

print(len(tokens))
print(tokens)
```

#### Segmentação de sentenças

Também podemos ter interesse em dividir o texto em sentenças.

```python
from nltk import sent_tokenize

text = "Hello. Are you Mr. Smith? Just to let you know that I have finished my M.Sc. and Ph.D. on AI. I loved it!"
sentences = sent_tokenize(text)

print(sentences)
print("Number of sentences:", len(sentences))
```

#### Experimentando com textos longos

Podemos tentar baixar um livro do [Project Gutenberg](https://www.gutenberg.org/).

```python
from urllib import request

url = "http://www.gutenberg.org/files/2554/2554-0.txt"
response = request.urlopen(url)
raw = response.read().decode('utf8')

print(len(raw))
print(raw[:75])
```

Quantas sentenças existem? Imprima a segunda sentença (índice 1).

```python
sentences = sent_tokenize(raw)

print(f'Number of sentences: {len(sentences)}')
print(f'Second sentence: {sentences[1]}')
```

Quantos tokens existem? Qual é o índice do primeiro token da segunda sentença?

```python
tokens = word_tokenize(raw)
tokens_second_sentence = word_tokenize(sentences[1])

print(len(tokens))
print(tokens_second_sentence)
```

E quantos types (tokens únicos) existem? Qual é o mais frequente? *(Dica: use um contêiner [Counter](https://docs.python.org/3/library/collections.html#collections.Counter) do módulo collections.)*

```python
import collections

counter = collections.Counter(tokens)

print(len(counter))
counter.most_common(1)
```

#### Lidando com expressões multipalavra (MWE)

Às vezes queremos que certas palavras permaneçam juntas na tokenization, como em nomes formados por várias palavras.

```python
word_tokenize("Good muffins cost $3.88\nin New York.")
```

Uma forma de fazer isso é fornecer nosso próprio léxico e usar o [tokenizer de MWE](https://www.nltk.org/api/nltk.tokenize.mwe.html) do NLTK.

```python
from nltk.tokenize import MWETokenizer
from nltk import sent_tokenize, word_tokenize

s = "Good muffins cost $3.88\nin New York."
mwe = MWETokenizer([('New', 'York'), ('Hong', 'Kong')], separator=' ')

[mwe.tokenize(word_tokenize(sent)) for sent in sent_tokenize(s)]
```

## Stemming e Lematização

*Stemming* e *lematização* são técnicas usadas para normalizar tokens, de modo a reduzir o tamanho do vocabulário.
Enquanto a lematização é o processo de encontrar a raiz (lema) da palavra, o stemming normalmente aplica um conjunto de regras de transformação que buscam cortar os afixos finais das palavras.

#### Stemming

O NLTK inclui um dos stemmers mais conhecidos: o [stemmer de Porter](https://www.emerald.com/insight/content/doi/10.1108/00330330610681286/full/pdf?casa_token=eT_IPtH_eLEAAAAA:Z3lAtxWdxf0FL479mL-A7tC-_QRzxNeeyC2DFLyWwGBlcj6DQcwu2Bnq37waDPcXKOnXkMMDtKGyCaYGZtYcb3lgBZ9uaHKUNO0JCMivSdPE4HTe).

```python
from nltk.stem import PorterStemmer

# initialize the Porter Stemmer
porter = PorterStemmer()
```

Vamos usar um trecho de texto ilustrativo:

```python
sentence = '''The European Commission has funded a numerical study to analyze the purchase of a pipe organ with no noise
for Europe's organization. Numerous donations have followed the analysis after a noisy debate.'''

# tokenize: split the text into words
word_list = nltk.word_tokenize(sentence)

print("\nOriginal word list:", word_list)
print("\nOriginal number of distinct tokens:", len(set(word_list)))
```

Agora aplicamos o stemming aos tokens do texto:

```python
# stem list of words and join
stemmed_output = ' '.join([porter.stem(w) for w in word_list])
print("Stemmed text:", stemmed_output)

# tokenize: split the text into words
stemmed_word_list = nltk.word_tokenize(stemmed_output)

print("\nStemmed word list:", stemmed_word_list)
print("\nStemmed number of distinct tokens:", len(set(stemmed_word_list)))
```

Você pode notar a redução no tamanho do vocabulário. Alguns tokens ficam supergeneralizados (tokens semanticamente diferentes que recebem o mesmo stem), enquanto outros ficam subgeneralizados (tokens semanticamente parecidos que recebem stems diferentes).

Experimente [outros stemmers](https://www.nltk.org/api/nltk.stem.html) disponíveis no NLTK.

```python
from nltk.stem import SnowballStemmer

stemmer = SnowballStemmer("english")

new_stemmed_output = ' '.join([stemmer.stem(w) for w in word_list])

new_stemmed_word_list = nltk.word_tokenize(new_stemmed_output)

print("\nStemmed word list:", new_stemmed_word_list)
print("\nStemmed number of distinct tokens:", len(set(new_stemmed_word_list)))
```

Podemos testar alguns para o português:

```python
# Portuguese stemmer: https://www.nltk.org/_modules/nltk/stem/rslp.html
from nltk.stem import RSLPStemmer

stemmer = RSLPStemmer()
sentence = "Estou mesmo a gostar desta unidade curricular, todos gostamos de unidades curriculares interessantes."

word_list = nltk.word_tokenize(sentence)
stemmed_output = ' '.join([stemmer.stem(w) for w in word_list])
print(stemmed_output)
```

#### Lematização

O NLTK inclui um [lematizador baseado no WordNet](https://www.nltk.org/api/nltk.stem.wordnet.html).

```python
# WordNet lemmatizer
from nltk.stem import WordNetLemmatizer 

# Init the Wordnet Lemmatizer
lemmatizer = WordNetLemmatizer()

sentence = "Men and women love to study artificial intelligence while studying data science. My feet and teeth are clean!"


# tokenize: Split the sentence into words
word_list = nltk.word_tokenize(sentence)
print(word_list)

# lemmatize list of words
lemmatized_output = [lemmatizer.lemmatize(w) for w in word_list]
print(lemmatized_output)
```

Compare o resultado com o stemming aplicado ao mesmo texto.

```python
stemmer = PorterStemmer()

stemmed_output = [stemmer.stem(w) for w in word_list]
print(stemmed_output)
```
